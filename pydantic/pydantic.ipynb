```python
from pydantic import BaseModel
```

```python
class person(BaseModel):
    name: str
    age: int
    email: str
    students: list[str] = []
    
person1 = person(name="John Doe", age=30, email="john.doe@example.com", students=["Alice", "Bob", "Charlie"])
print(person1)
```

```python

from typing import List
class classroom(BaseModel):
    name: str
    students: List[person] = []
    capacity: int = 30
```

```python
classroom=classroom(name="Math 101", students=[person1], capacity=30)
```
